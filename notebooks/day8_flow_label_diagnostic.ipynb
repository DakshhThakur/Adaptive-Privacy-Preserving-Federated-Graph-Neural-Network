```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
DATA_PATH = Path("../data/processed/graph_metadata.csv")

metadata = pd.read_csv(DATA_PATH)

print("Shape:", metadata.shape)
print("\nColumns:")
print(metadata.columns.tolist())

metadata.head()
```

```python
print("Flow-level label distribution:")
print(metadata["Binary_Label"].value_counts())

print("\nPercentage:")
print(metadata["Binary_Label"].value_counts(normalize=True) * 100)
```

```python
print("Original Label distribution:")
print(metadata["Label"].value_counts())
```

```python
attack_flows = metadata[metadata["Binary_Label"] == 1]

print("Total attack flows:", len(attack_flows))
print("Total benign flows:", len(metadata[metadata["Binary_Label"] == 0]))
```

```python
unique_attack_sources = attack_flows["Source IP"].nunique()

print("Unique source IPs involved in attacks:", unique_attack_sources)
```

```python
unique_attack_destinations = attack_flows["Destination IP"].nunique()

print("Unique destination IPs involved in attacks:", unique_attack_destinations)
```

```python
print("Attack types:")

print(
    attack_flows["Label"]
    .value_counts()
)
```

```python
print("========== FLOW-LEVEL DIAGNOSTIC ==========")

print("Total flows:", len(metadata))
print("Benign flows:", (metadata["Binary_Label"] == 0).sum())
print("Attack flows:", (metadata["Binary_Label"] == 1).sum())

print(
    "Unique attack source IPs:",
    metadata.loc[
        metadata["Binary_Label"] == 1,
        "Source IP"
    ].nunique()
)

print(
    "Unique attack destination IPs:",
    metadata.loc[
        metadata["Binary_Label"] == 1,
        "Destination IP"
    ].nunique()
)

print("\nAttack types:")
print(
    metadata.loc[
        metadata["Binary_Label"] == 1,
        "Label"
    ].value_counts()
)
```
